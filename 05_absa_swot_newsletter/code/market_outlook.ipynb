```python
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o", temperature=0.7)

# 향후 전망을 위한 프롬프트 구성
template = """
{article_summary}
위 요약을 기반으로 향후 시장 전망을 제시해줘. 기회와 위협 모두 고려할 것.
"""

chain = LLMChain(llm=llm, prompt=PromptTemplate(input_variables=["article_summary"], template=template))

def generate_outlook(summary):
    """
    뉴스 요약 기반으로 향후 시장에 대한 전망을 도출합니다.
    """
    return chain.invoke({"article_summary": summary}).content
```
