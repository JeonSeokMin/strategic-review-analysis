```python
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from langchain_openai import ChatOpenAI

# LLM 초기화
llm = ChatOpenAI(temperature=0.7, model_name="gpt-4o")

# 감정별 프롬프트 생성 함수
def build_prompt(polarity: str):
    """
    감정 극성(polarity)에 따라 ABSA 분석용 프롬프트를 생성합니다.
    """
    template = f"""
    Define the sentiment elements...

    − The 'sentiment polarity' expresses whether the sentiment is '{polarity}'.

    Provide two {polarity} sentiment examples:
    {{text}}
    """
    return PromptTemplate(template=template, input_variables=["text"])

# 각 감정 극성별 체인 구성
chains = {
    "positive": LLMChain(llm=llm, prompt=build_prompt("positive")),
    "negative": LLMChain(llm=llm, prompt=build_prompt("negative")),
    "neutral": LLMChain(llm=llm, prompt=build_prompt("neutral")),
}

def run_absa_analysis(text):
    """
    입력된 리뷰 텍스트를 ABSA 분석하여
    긍정 / 부정 / 중립 결과를 반환합니다.
    """
    return {
        "positive": chains["positive"].invoke({"text": text}).content,
        "negative": chains["negative"].invoke({"text": text}).content,
        "neutral": chains["neutral"].invoke({"text": text}).content,
    }
```
