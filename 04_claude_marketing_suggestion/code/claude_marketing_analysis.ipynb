```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

import anthropic
```

```python
predict_result = pd.read_csv('../../01_kobert_finetuning_prediction/results/kobert_raw_predictions.csv')
```

```python
print("긍정 리뷰: ", len(predict_result[predict_result['AI_HUB_PT_predict'] == 1]['리뷰상세내용']))
print("부정 리뷰: ", len(predict_result[predict_result['AI_HUB_PT_predict'] == -1]['리뷰상세내용']))
```

```python
# 1. 데이터 준비
negative_reviews = predict_result[predict_result['AI_HUB_PT_predict'] == -1]['리뷰상세내용'].tolist()

# 2. 데이터를 문자열로 변환
dataset_str = "\n".join([f"{i+1}. {review}" for i, review in enumerate(negative_reviews)])


# 3. Claude API 호출
client = anthropic.Anthropic(
    # api_key = my_api
)

response = client.messages.create(
    model="claude-3-5-sonnet-20240620",
    max_tokens=2048,
    system="You are a seasoned data scientist at a Fortune 500 company specializing in consumer product analysis and strategic decision-making. Your task is to analyze negative customer reviews and provide strategic recommendations to improve product satisfaction and address customer concerns.",
    messages=[
        {"role": "user", "content": f"""Analyze this dataset of negative customer reviews for a health supplement product and provide strategic recommendations:

<dataset>
{dataset_str}
</dataset>

Please provide:
1. A summary of the main issues identified in the reviews.
2. Key insights derived from the analysis.
3. Strategic recommendations for the company to address these issues and improve customer satisfaction.
4. Potential areas for further research or data collection."""}
    ]
)

# 4. 결과 출력
print(response.content)
```
