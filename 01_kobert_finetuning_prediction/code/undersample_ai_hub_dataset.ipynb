```python
import pandas as pd
```

## 'AI_HUB 속성기반 감정분석 데이터셋' 호출 후 라벨 값 확인

```python
dataset = pd.read_csv('../data/ai_hub_raw.csv')

class_counts = dataset['GeneralPolarity'].value_counts()

print(class_counts)
```

## 언더샘플링 진행

```python
import pandas as pd
import random

positive_indices = dataset[dataset['GeneralPolarity'] == 1.0].index.tolist()

num_positive_samples = len(dataset[dataset['GeneralPolarity'] == 1.0])

undersampled_positive_indices = random.sample(positive_indices, num_positive_samples - 26367)
undersampled_dataset = dataset.drop(undersampled_positive_indices)

print(undersampled_dataset['GeneralPolarity'].value_counts())
```

## 언더샘플링 데이터셋 별도 저장

```python
undersampled_dataset.to_csv('aihub_undersampled_data.csv', index=False)
```
